This notebook demonstrates how to perform simple linear regression analysis to understand the relationship between education levels and crime rates.

## Step 2: Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

print("All libraries imported successfully!")
```

## Step 3: Read and Describe the Data

```python
# Read the CSV file
df = pd.read_csv("simpledata.csv")

# Display first few rows
print("First 10 rows of the dataset:")
print(df.head(10))
print("\n" + "="*50 + "\n")

# Display dataset information
print("Dataset Information:")
print(f"Number of observations: {len(df)}")
print(f"Number of variables: {len(df.columns)}")
print("\n" + "="*50 + "\n")

# Statistical description
print("Statistical Summary:")
print(df.describe())
```

## Step 4: Separate Independent (X) and Dependent (Y) Variables

```python
# Define feature and target columns
features_col = ["Education"]
target = ["Crime"]

# Extract X (independent variable) and Y (dependent variable)
x = np.array(df[features_col])[:, 0]
y = np.array(df[target])[:, 0]

print("Independent Variable (X - Education):")
print(x)
print("\n" + "="*50 + "\n")

print("Dependent Variable (Y - Crime):")
print(y)
print("\n" + "="*50 + "\n")

print(f"Shape of X: {x.shape}")
print(f"Shape of Y: {y.shape}")
```

## Step 5: Visualize the Data (Scatter Plot)

```python
# Create scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(x, y, color='blue', alpha=0.6, s=50, label='Data Points')
plt.xlabel('Education (years)', fontsize=12)
plt.ylabel('Crime Rate', fontsize=12)
plt.title('Education vs Crime Rate - Scatter Plot', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## Step 6: Apply Linear Regression

```python
# Reshape X for sklearn (needs 2D array)
X_reshaped = x.reshape(-1, 1)

# Create and fit the linear regression model
model = LinearRegression()
model.fit(X_reshaped, y)

# Make predictions
y_pred = model.predict(X_reshaped)

print("Linear Regression Model fitted successfully!")
print(f"Number of training samples: {len(X_reshaped)}")
```

## Step 7: Print Linear Regression Parameters

```python
# Extract model parameters
slope = model.coef_[0]
intercept = model.intercept_

# Calculate performance metrics
r2 = r2_score(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)

# Calculate correlation coefficient
correlation = np.corrcoef(x, y)[0, 1]

print("="*60)
print("LINEAR REGRESSION RESULTS")
print("="*60)
print(f"\nRegression Equation: Crime = {intercept:.2f} + ({slope:.2f}) × Education")
print("\n" + "-"*60)
print("\nModel Parameters:")
print(f"  Slope (β₁):              {slope:.4f}")
print(f"  Intercept (β₀):          {intercept:.4f}")
print("\n" + "-"*60)
print("\nModel Performance Metrics:")
print(f"  R-squared (R²):          {r2:.4f}")
print(f"  Correlation (r):         {correlation:.4f}")
print(f"  Mean Squared Error:      {mse:.2f}")
print(f"  Root Mean Squared Error: {rmse:.2f}")
print("="*60)
```

## Step 8: Plot Regression Line with Data Points

```python
# Create comprehensive plot
plt.figure(figsize=(12, 7))

# Plot actual data points
plt.scatter(x, y, color='blue', alpha=0.6, s=60, label='Actual Data', zorder=3)

# Plot regression line
plt.plot(x, y_pred, color='red', linewidth=2.5, label='Regression Line', zorder=2)

# Add labels and title
plt.xlabel('Education (years)', fontsize=13)
plt.ylabel('Crime Rate', fontsize=13)
plt.title('Linear Regression: Education vs Crime Rate', fontsize=15, fontweight='bold')

# Add regression equation to plot
equation_text = f'Crime = {intercept:.2f} + ({slope:.2f}) × Education\nR² = {r2:.4f}'
plt.text(0.05, 0.95, equation_text, transform=plt.gca().transAxes, 
         fontsize=11, verticalalignment='top', 
         bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

plt.grid(True, alpha=0.3, linestyle='--')
plt.legend(loc='upper right', fontsize=11)
plt.tight_layout()
plt.show()
```

## Step 9: Residual Analysis

```python
# Calculate residuals
residuals = y - y_pred

# Create residual plots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Residual plot
axes[0].scatter(y_pred, residuals, color='purple', alpha=0.6)
axes[0].axhline(y=0, color='red', linestyle='--', linewidth=2)
axes[0].set_xlabel('Predicted Values', fontsize=11)
axes[0].set_ylabel('Residuals', fontsize=11)
axes[0].set_title('Residual Plot', fontsize=12, fontweight='bold')
axes[0].grid(True, alpha=0.3)

# Histogram of residuals
axes[1].hist(residuals, bins=15, color='green', alpha=0.7, edgecolor='black')
axes[1].set_xlabel('Residuals', fontsize=11)
axes[1].set_ylabel('Frequency', fontsize=11)
axes[1].set_title('Distribution of Residuals', fontsize=12, fontweight='bold')
axes[1].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

print(f"Mean of Residuals: {np.mean(residuals):.4f}")
print(f"Standard Deviation of Residuals: {np.std(residuals):.4f}")
```

## Step 10: Analysis and Conclusions

```python
print("="*70)
print("ANALYSIS AND CONCLUSIONS")
print("="*70)

print("\n1. RELATIONSHIP BETWEEN VARIABLES:")
if slope < 0:
    print(f"   - There is a NEGATIVE relationship between education and crime.")
    print(f"   - For every 1-year increase in education, crime rate decreases")
    print(f"     by approximately {abs(slope):.2f} units.")
else:
    print(f"   - There is a POSITIVE relationship between education and crime.")
    print(f"   - For every 1-year increase in education, crime rate increases")
    print(f"     by approximately {slope:.2f} units.")

print("\n2. STRENGTH OF RELATIONSHIP:")
if abs(correlation) > 0.7:
    strength = "STRONG"
elif abs(correlation) > 0.4:
    strength = "MODERATE"
else:
    strength = "WEAK"
print(f"   - Correlation coefficient (r) = {correlation:.4f}")
print(f"   - This indicates a {strength} {'negative' if correlation < 0 else 'positive'} correlation.")

print("\n3. MODEL PERFORMANCE:")
print(f"   - R-squared (R²) = {r2:.4f}")
print(f"   - This means that {r2*100:.2f}% of the variation in crime rate")
print(f"     can be explained by education level.")

if r2 > 0.7:
    fit_quality = "EXCELLENT"
elif r2 > 0.5:
    fit_quality = "GOOD"
elif r2 > 0.3:
    fit_quality = "MODERATE"
else:
    fit_quality = "POOR"
print(f"   - Model fit quality: {fit_quality}")

print("\n4. PREDICTION ACCURACY:")
print(f"   - Root Mean Squared Error (RMSE) = {rmse:.2f}")
print(f"   - On average, predictions deviate from actual values by {rmse:.2f} units.")

print("\n5. PRACTICAL IMPLICATIONS:")
if slope < 0:
    print("   - Higher education levels are associated with lower crime rates.")
    print("   - Investment in education may help reduce crime in communities.")
else:
    print("   - Higher education levels are associated with higher crime rates.")
    print("   - This unexpected result may require further investigation.")

print("\n6. LIMITATIONS:")
print("   - This is a simple linear model with only one predictor variable.")
print("   - Other factors (poverty, employment, demographics) also affect crime.")
print("   - Correlation does not imply causation.")

print("\n" + "="*70)
```

## Step 11: Make Predictions

```python
# Example predictions for different education levels
education_levels = np.array([6, 8, 10, 12, 14]).reshape(-1, 1)
predicted_crime = model.predict(education_levels)

print("PREDICTIONS FOR DIFFERENT EDUCATION LEVELS:")
print("="*50)
for edu, crime in zip(education_levels.flatten(), predicted_crime):
    print(f"Education: {edu:4.1f} years → Predicted Crime Rate: {crime:7.2f}")
print("="*50)
```
